```python
N = 4
board = [[0] * N for _ in range(N)]

def is_safe(r, c):
    for i in range(r):

        if board[i][c] == 1 or \
           (c - r + i >= 0 and board[i][c - r + i] == 1) or \
           (c + r - i < N and board[i][c + r - i] == 1):
            return False
    return True

def solve(r):

    if r == N:
        for row in board:
            print(row)
        return True


    for c in range(N):
        if is_safe(r, c):
            board[r][c] = 1

            if solve(r + 1):
                return True

            board[r][c] = 0

    return False

solve(0)
```

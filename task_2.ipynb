```python
def objective_function(x):
    return -(x - 5) ** 2 + 25


def hill_climbing(current_state, step=1, stop_criterion=10):
    current_value = objective_function(current_state)

    print("Initial state:", current_state)
    print("Initial objective value:", current_value)

    for iteration in range(1, stop_criterion + 1):
        next_state = current_state + step
        next_value = objective_function(next_state)

        print(f"\nIteration {iteration}")
        print("Changed state:", next_state)
        print("Objective value:", next_value)

        if next_value > current_value:
            current_state = next_state
            current_value = next_value

            print("Objective function is greater")
            print("New current state:", current_state)

        else:
            print("Objective function is not greater")

            if current_state == 5:
                print("\nStop criterion reached")
                break

    return current_state, current_value


initial_state = 0
best_state, best_value = hill_climbing(initial_state)

print("\nFinal result")
print("Best state:", best_state)
print("Best objective value:", best_value)
```
